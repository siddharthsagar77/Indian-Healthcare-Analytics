```python

for col in df .columns:
    if df[col].nunique() < 10:
        print(df[col].value_counts())
        print('--------------------------------')
```

```python
# dropping 2 columns Imaging_Type,Imaging_Findings
df.dropna(subset=['Imaging_Type', 'Imaging_Findings'], inplace=True)
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
# Age group segmentation
bins  = [0, 12, 17, 35, 60, 100]
labels = ["Child (0-12)","Teen (13-17)","Young Adult (18-35)",
          "Middle-Aged (36-60)","Senior (60+)"]
df["Age_Group"] = pd.cut(df["Age"], bins=bins, labels=labels)
```

```python
# BMI Category
def bmi_cat(bmi):
    if bmi < 18.5: return "Underweight"
    elif bmi < 25:  return "Normal"
    elif bmi < 30:  return "Overweight"
    else:           return "Obese"
df["BMI_Category"] = df["BMI"].apply(bmi_cat)
```

```python
df.columns
```

```python
df["Insurance_Int"] = df["Insurance_Covered"].astype(int)
```

```python
df.columns
```

```python
df['Visit_Date'] = pd.to_datetime(df['Visit_Date'])
```

```python
df.info()
```
